```python
import pandas as pd
import numpy as np
import pickle
```

# 2. Load Model

```python
# Load model Random Forest
with open('P1M2_Angelina_Yessyca.pkl', 'rb') as file_1:
  model_rf = pickle.load(file_1)

# Load Winsorizer
with open('windsoriser.pkl', 'rb') as file_2:
  windsoriser = pickle.load(file_2)

# Load Label Encoder
with open('label_encoder.pkl', 'rb') as file_3:
  label_encoder = pickle.load(file_3)
```

# 3. Membuat Data Inference

```python
# Membuat 1 data baru
data_inf = {
    'age' : 39,
    'job' : 'entrepreneur',
    'marital' : 'single',
    'education' : 'basic.9y',
    'default' : 'no',
    'housing' : 'yes',
    'loan' : 'no',
    'contact' : 'cellular',
    'month' : 'dec',
    'day_of_week' : 'fri',
    'duration' : 390,
    'campaign' : 3,
    'pdays' : 999,
    'previous' : 0,
    'poutcome' : 'nonexistent',
    'emp.var.rate' : -1.8,
    'cons.price.idx' : 92.893,
    'cons.conf.idx' : -46.2,
    'euribor3m' : 1.313,
    'nr.employed' : 5099.1
}

# Mengubah dictionary menjadi DataFrame
data_inf = pd.DataFrame([data_inf])
data_inf
```

# 4. Preprocessing Data Inference

```python
# Mengubah nilai 999 pada pdays menjadi missing value
data_inf['pdays'] = data_inf['pdays'].replace(999, np.nan)

# Menangani outlier menggunakan winsoriser yang sudah dilatih
data_inf = windsoriser.transform(data_inf)

# Menghapus fitur duration
data_inf = data_inf.drop(['duration'], axis = 1)
data_inf
```

# 5. Melakukan Prediksi

```python
# Melakukan prediksi
y_pred_inf = model_rf.predict(data_inf)
y_pred_inf
```
